This notebook:
1. Loads trained PyTorch checkpoints for ISL and ASL.
2. Exports dynamic-batch ONNX graphs.
3. Validates numerical parity between PyTorch and ONNX Runtime ($< 10^{-4}$ tolerance).
4. Benchmarks inference latency (ms per sequence).
5. Generates browser-side metadata JSON files and prepares artifacts for deployment.

```python
# Step 2: Export ISL Model to ONNX
from ml.signbridge.export import export_to_onnx

try:
    isl_res = export_to_onnx(
        checkpoint_path='checkpoints/isl/best_model.pt',
        output_onnx_path='public/models/isl-bilstm-v1.onnx',
        metadata_output_path='public/models/isl-metadata.json'
    )
    print('ISL Export Success:', isl_res)
except Exception as e:
    print('Train ISL model in Notebook 4 before export:', e)
```

```python
# Step 3: Export ASL Model to ONNX
try:
    asl_res = export_to_onnx(
        checkpoint_path='checkpoints/asl/best_model.pt',
        output_onnx_path='public/models/asl-bilstm-v1.onnx',
        metadata_output_path='public/models/asl-metadata.json'
    )
    print('ASL Export Success:', asl_res)
except Exception as e:
    print('Train ASL model in Notebook 4 before export:', e)
```
