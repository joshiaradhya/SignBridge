This notebook:
1. Loads dataset manifests for ISL and ASL.
2. Executes batch feature extraction using **MediaPipe Hands, Upper-Pose, and Face Mesh**.
3. Normalizes 74 3D landmarks (222 features per frame) using torso/wrist-relative normalization.
4. Resamples sequences to 32 frames.
5. Serializes extracted datasets to compressed `.npz` files respecting signer splits.

```python
# Step 2: Extract ISL landmarks from INCLUDE-50 manifest
from pathlib import Path
from ml.signbridge.datasets import DatasetManifest
from ml.signbridge.extraction import process_manifest_to_npz

isl_manifest_path = Path('data/manifests/include50_manifest.json')
if isl_manifest_path.exists():
    manifest = DatasetManifest.load(isl_manifest_path)
    stats = process_manifest_to_npz(manifest, 'data/isl_features_v2.npz')
    print('ISL extraction finished:', stats)
    !cp data/isl_features_v2.npz /content/drive/MyDrive/SignBridge_Data/ 2>/dev/null || true
else:
    print('Generate manifest in Notebook 1 first.')
```

```python
# Step 3: Extract ASL landmarks from ASL Citizen manifest
asl_manifest_path = Path('data/manifests/asl_citizen_manifest.json')
if asl_manifest_path.exists():
    manifest = DatasetManifest.load(asl_manifest_path)
    stats = process_manifest_to_npz(manifest, 'data/asl_features_v2.npz')
    print('ASL extraction finished:', stats)
    !cp data/asl_features_v2.npz /content/drive/MyDrive/SignBridge_Data/ 2>/dev/null || true
else:
    print('Generate manifest in Notebook 1 first.')
```
